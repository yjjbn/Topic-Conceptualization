# 50 runs comparison

```python
models = ("openai", "google_gemini-3.5", "moonshot", "mistral")
order_by = "openai"

all_df = pl.concat(
    [
        pl.read_csv(file).with_columns(
            pl.lit(file.parent.parent.name).alias("run"),
            pl.lit(file.parent.name).alias("model"),
            pl.lit(file.name).alias("source"),
        )
        for file in sorted(
            Path("../01_codebook_dev/out/grimmer_50").rglob("classifications*.csv")
        )
        if file.parent.name.startswith(models)
    ],
    how="diagonal_relaxed",
)

# Calculate agreement separately for each model and text.
all_counts = (
    all_df.group_by("document_id", "text", "model")
    .agg(
        (pl.col("label").cast(pl.String) == "0").sum().alias("n_0"),
        (pl.col("label").cast(pl.String) == "1").sum().alias("n_1"),
    )
    .with_columns(
        pl.max_horizontal("n_0", "n_1").alias("n_agree")
    )
    .with_columns(
        (100 * pl.col("n_agree") / (pl.col("n_0") + pl.col("n_1")))
        .alias("agreement_pct")
    )
)

# Selected model: majority-1 texts (most 1s first), then majority-0 texts (most 0s first).
# Exact ties come last.
text_order = (
    all_counts.filter(pl.col("model").str.starts_with(order_by))
    .with_columns(
        pl.when(pl.col("n_1") > pl.col("n_0")).then(0)
        .when(pl.col("n_0") > pl.col("n_1")).then(1)
        .otherwise(2).alias("label_group")
    )
    .sort(["label_group", "n_agree", "document_id"], descending=[False, True, False])
    ["document_id"].to_list()
)
matrix_order = text_order

matrices = {}
counts_by_model = {}
Path("./out").mkdir(exist_ok=True)

for mod in models:
    df = all_df.filter(pl.col("model").str.starts_with(mod))
    counts = all_counts.filter(pl.col("model").str.starts_with(mod))
    counts_by_model[mod] = counts.to_pandas().set_index("document_id").reindex(text_order)
    matrices[mod] = df.to_pandas().pivot(
        index="document_id", columns=["model", "run"], values="label"
    ).reindex(matrix_order)
    counts.sort("agreement_pct").write_csv(f"./out/most_disagree_grimmer_{mod}.csv")

# Agreement charts side by side.
fig, axes = plt.subplots(1, len(models), figsize=(12, 5), sharey=True)
for ax, mod in zip(axes, models):
    counts_by_model[mod].plot.bar(y="agreement_pct", ax=ax, legend=False)
    ax.set(xlabel="Texts", ylabel="% Agreement", title=mod, ylim=(0, 100))
    ax.axhline(y=90, color="red")
plt.show()

# Classification matrices side by side, with identical text order and colors.
fig, axes = plt.subplots(1, len(models), figsize=(12, 5), sharey=True)
for ax, mod in zip(axes, models):
    ax.imshow(matrices[mod].astype(float), aspect="auto", cmap="Greys", vmin=0, vmax=1)
    ax.set(xlabel="Model runs", ylabel="Texts", title=mod)
plt.show()
```

```python
pl.DataFrame([
    {
        "model": mod,
        "alpha": krippendorff.alpha(
            reliability_data=matrices[mod].astype(float).to_numpy().T,
            level_of_measurement="nominal",
        ),
        "pct_texts_100": (
            100 * (counts_by_model[mod]["agreement_pct"] == 100).mean()
        ),
        "pct_texts_above_90": (
            100 * (counts_by_model[mod]["agreement_pct"] > 90).mean()
        ),
        "pct_texts_above_75": (
            100 * (counts_by_model[mod]["agreement_pct"] > 80).mean()
        ),
    }
    for mod in models
])
```

```python
results = []
expected_models = all_df["model"].n_unique()

for (run,), group in all_df.group_by("run"):
    wide = group.pivot(
        index="document_id", on="model", values="label"
    ).drop("document_id").drop_nulls()

    pct_agreement = None
    pct_at_least_3_of_4 = None
    alpha = None

    # Compare only runs containing all selected models.
    if wide.width == expected_models and wide.height and expected_models >= 2:
        ratings = wide.cast(pl.Float64).to_numpy().T

        pct_agreement = 100 * (ratings == ratings[0]).all(axis=0).mean()
        if wide.width == 4:
            votes_1 = (ratings == 1).sum(axis=0)
            votes_0 = (ratings == 0).sum(axis=0)
            pct_at_least_3_of_4 = 100 * ((votes_1 >= 3) | (votes_0 >= 3)).mean()

        if ratings.min() != ratings.max():
            alpha = krippendorff.alpha(
                reliability_data=ratings,
                level_of_measurement="nominal",
            )

    results.append({
        "run": int(run.removeprefix("run_")),
        "n_models": wide.width,
        "agreement_pct": pct_agreement,
        "pct_texts_at_least_3_of_4": pct_at_least_3_of_4,
        "alpha": alpha,
    })

within_run = pl.DataFrame(results).sort("run")

within_run.select(
    pl.col("agreement_pct").mean().alias("mean_agreement_pct"),
    pl.col("alpha").mean().alias("mean_alpha"),
    pl.col("pct_texts_at_least_3_of_4").mean().alias("mean_pct_texts_at_least_3_of_4"),
)
```

# Validate codebooks

```python
orig = pl.concat(
    [
        pl.read_csv(file).with_columns(
            pl.lit(file.name).alias("source")
        )
        for file in sorted(
            Path("../01_codebook_dev/out/explanation_codebookLLM/run_1").rglob("classifications*.csv")
        )
    ],
    how="diagonal_relaxed",
)

orig_wide = (
    orig.with_columns(
        pl.col("source")
        .str.strip_prefix("classifications_explanation_codebookLLM_")
        .str.strip_suffix(".csv")
        # .str.replace_all("redo_", "", literal=True)
        .alias("model_run")
    )
    .pivot(index="document_id", on="model_run", values="label")
    .sort("document_id")
)

new = pl.concat(
    [
        pl.read_csv(file).with_columns(
            pl.lit(file.name).alias("source"),
            pl.lit(file.parent.as_posix())
            .str.extract(r"(run_\d)", 1)
            .alias("run"),
        )
        for file in sorted(
            Path("../02_experiment/out/explanation_codebookLLM_reapply").rglob("results*.csv")
        )
    ],
    how="diagonal_relaxed",
)

new_wide = (
    new.with_columns(
        pl.concat_str(
            pl.col("run"),
            pl.col("source")
            .str.strip_prefix("results_")
            .str.strip_suffix(".csv"),
            # .str.replace_all("_reapply_cbllm_redo", "", literal=True),
            separator="",
        ).alias("model_run")
    )
    .pivot(index="document_id", on="model_run", values="label")
    .sort("document_id")
)
new
```

```python
models = [
    col for col in orig_wide.columns
    if col != "document_id" and col in new_wide.columns
]

results = []

for model in models:
    paired = (
        orig_wide.select(
            pl.col("document_id").cast(pl.String),
            pl.col(model).cast(pl.String).alias("actual"),
        )
        .join(
            new_wide.select(
                pl.col("document_id").cast(pl.String),
                pl.col(model).cast(pl.String).alias("predicted"),
            ),
            on="document_id",
            how="inner",
            validate="1:1",
        )
        .drop_nulls(["actual", "predicted"])
    )

    actual = paired["actual"]
    predicted = paired["predicted"]

    tp = ((actual == "1") & (predicted == "1")).sum()
    fp = ((actual == "0") & (predicted == "1")).sum()
    fn = ((actual == "1") & (predicted == "0")).sum()

    # Alpha is undefined without observations or variation in the labels.
    alpha = None
    if paired.height and pl.concat([actual, predicted]).n_unique() > 1:
        alpha = krippendorff.alpha(
            reliability_data=(
                paired.select("actual", "predicted")
                .cast(pl.Float64)
                .to_numpy()
                .T
            ),
            level_of_measurement="nominal",
        )

    results.append({
        "model": model,
        "n_compared": paired.height,
        "accuracy": (
            (actual == predicted).mean() if paired.height else None
        ),
        "precision": tp / (tp + fp) if tp + fp else None,
        "recall": tp / (tp + fn) if tp + fn else None,
        "f1": 2 * tp / (2 * tp + fp + fn) if 2 * tp + fp + fn else None,
        "krippendorff_alpha": alpha,
    })

metrics = pl.DataFrame(results)
metrics
```

## Schub 4 x 4: agreement within codebooks and within models
Each table uses texts with all four valid labels. The 3-of-4 percentage includes 4-of-4 agreement. Alpha uses the same complete texts; it is null without label variation.

```python
from pathlib import Path
import polars as pl
import krippendorff

cross_root = Path("../02_experiment/out/schub_4x4")
application_run = 1
source_run = 1

cross_files = sorted(cross_root.rglob("results_*.csv"))
if not cross_files:
    raise ValueError(f"No retrieved results under {cross_root}")

cross_results = pl.concat(
    [pl.read_csv(file, schema_overrides={
        "document_id": pl.String,
        "label": pl.String,
        "_retrieval_error": pl.String,
    }) for file in cross_files],
    how="diagonal_relaxed",
).filter(
    (pl.col("application_run") == application_run)
    & (pl.col("source_run") == source_run)
)

cross_keys = ["application_model", "source_model", "document_id"]
if cross_results.unique(subset=cross_keys).height != cross_results.height:
    raise ValueError("Multiple batches cover the same combination. Filter to the intended batch_id values first.")

# Invalid or missing classifications are excluded from complete four-way comparisons.
cross_results = cross_results.with_columns(
    pl.when(pl.col("label").is_in(["0", "1"]) & pl.col("_retrieval_error").is_null())
    .then(pl.col("label"))
    .otherwise(None)
    .alias("label")
)


def four_way_agreement(group, rater_column):
    wide = group.pivot(index="document_id", on=rater_column, values="label").drop("document_id")
    complete = wide.drop_nulls() if wide.width == 4 else wide.head(0)
    metrics = {
        "n_raters": wide.width,
        "n_texts": wide.height,
        "n_compared": complete.height,
        "n_excluded": wide.height - complete.height,
        "pct_agree_4_of_4": None,
        "pct_agree_at_least_3_of_4": None,
        "krippendorff_alpha": None,
    }
    if complete.height:
        ratings = complete.cast(pl.Float64).to_numpy().T
        votes_1 = (ratings == 1).sum(axis=0)
        votes_0 = (ratings == 0).sum(axis=0)
        metrics["pct_agree_4_of_4"] = 100 * ((votes_1 == 4) | (votes_0 == 4)).mean()
        metrics["pct_agree_at_least_3_of_4"] = 100 * ((votes_1 >= 3) | (votes_0 >= 3)).mean()
        if ratings.min() != ratings.max():
            metrics["krippendorff_alpha"] = krippendorff.alpha(
                reliability_data=ratings,
                level_of_measurement="nominal",
            )
    return metrics
```

```python
# Hold the codebook fixed; compare the four application models.
within_codebooks_4x4 = pl.DataFrame([
    {
        "codebook_model": codebook_model,
        **four_way_agreement(group, "application_model"),
    }
    for (codebook_model,), group in cross_results.group_by("source_model")
]).sort("codebook_model")
within_codebooks_4x4
```

```python
# Hold the application model fixed; compare its labels under the four codebooks.
within_models_4x4 = pl.DataFrame([
    {
        "application_model": application_model,
        **four_way_agreement(group, "source_model"),
    }
    for (application_model,), group in cross_results.group_by("application_model")
]).sort("application_model")
within_models_4x4
```

```python
df = pl.concat(
    [
        pl.read_csv(file).with_columns(
            pl.lit(file.name).alias("model"),
            pl.lit(
                next(
                    (parent.name for parent in file.parents
                     if "codebook" in parent.name),
                    None,
                )
            ).alias("codebook"),
        )
        for file in sorted(
            Path("../02_experiment/out/explanation_codebookLLM/run_1").rglob("results*.csv")
        )
    ],
    how="diagonal_relaxed",
)

majority = (
    df.with_columns(pl.col("label").cast(pl.Int64))
    .group_by(["model", "codebook", "document_id"])
    .agg(
        (pl.col("label") == 1).sum().alias("votes_1"),
        (pl.col("label") == 0).sum().alias("votes_0"),
    )
    .with_columns(
        pl.when(pl.col("votes_1") > pl.col("votes_0"))
        .then(pl.lit(1))
        .when(pl.col("votes_0") > pl.col("votes_1"))
        .then(pl.lit(0))
        .otherwise(None)
        .alias("label")
    )
    .sort(["model", "codebook", "document_id"])
)

def agreement(df: pl.DataFrame) -> dict:

    ratings = df.cast(pl.Int64).to_numpy()
    unanimous = (ratings.sum(axis=1) == 0) | (ratings.sum(axis=1) == df.width)

    return {
        "percent_agreement": 100 * unanimous.mean(),
        "krippendorff_alpha": (
            float(krippendorff.alpha(
                reliability_data=ratings.T,
                level_of_measurement="nominal",
            ))
            if ratings.min() != ratings.max() else None
        ),
    }

majority = majority.filter(~pl.col("model").str.contains("mistral")).filter(~pl.col("codebook").str.contains("mistral"))

dfs = {
    "model_google": majority.filter(pl.col("model").str.contains("google")).pivot(index="document_id", on="codebook", values="label"),
    # "model_mistral": majority.filter(pl.col("model").str.contains("mistral")).pivot(index="document_id", on="codebook", values="label"),
    "model_terra": majority.filter(pl.col("model").str.contains("terra")).pivot(index="document_id", on="codebook", values="label"),
    "cb_google": majority.filter(pl.col("codebook").str.contains("google")).pivot(index="document_id", on="model", values="label"),
    # "cb_mistral": majority.filter(pl.col("codebook").str.contains("mistral")).pivot(index="document_id", on="model", values="label"),
    "cb_terra": majority.filter(pl.col("codebook").str.contains("terra")).pivot(index="document_id", on="model", values="label"),
}

pl.DataFrame([
    {"comparison": name, **agreement(frame.drop("document_id"))}
    for name, frame in dfs.items()
])
```

## Original versus reapplied codebook classifications
Use the output nametag from apply_codebook_batch.py. Each row is paired with its original model/run/document. Missing application results stay missing, rather than counting as disagreement.

```python
reapply_root = Path("../02_experiment/out/grimmer_50_reapply")
result_files = sorted(reapply_root.rglob("results_*.csv"))
if not result_files:
    raise ValueError(f"No retrieved results under {reapply_root}")

paired = pl.concat(
    [pl.read_csv(file, schema_overrides={
        "document_id": pl.String,
        "original_label": pl.String,
        "reapplied_label": pl.String,
        "agrees_with_original": pl.Boolean,
        "source_run": pl.Int64,
    }) for file in result_files],
    how="diagonal_relaxed",
)

reapply_agreement = (
    paired.group_by("source_model", "source_run")
    .agg(
        pl.len().alias("n_texts"),
        pl.col("reapplied_label").is_null().sum().alias("n_missing_reapplied"),
        pl.col("original_label").is_null().sum().alias("n_missing_original"),
        pl.col("agrees_with_original").count().alias("n_compared"),
        pl.col("agrees_with_original").sum().alias("n_agree"),
        (100 * pl.col("agrees_with_original").mean()).alias("agreement_pct"),
    )
    .sort("source_model", "source_run")
)
# Treat original and reapplied classifications as two nominal raters.
alpha_rows = []
for (model, run), group in paired.group_by("source_model", "source_run"):
    ratings = (
        group.select("original_label", "reapplied_label")
        .drop_nulls()
        .cast(pl.Float64)
        .to_numpy()
        .T
    )
    # Alpha is undefined if no complete pairs or only one label is observed.
    alpha = None
    if ratings.size and ratings.min() != ratings.max():
        alpha = krippendorff.alpha(
            reliability_data=ratings,
            level_of_measurement="nominal",
        )
    alpha_rows.append({
        "source_model": model,
        "source_run": run,
        "krippendorff_alpha": alpha,
    })

reapply_agreement = reapply_agreement.join(
    pl.DataFrame(alpha_rows, schema_overrides={"krippendorff_alpha": pl.Float64}),
    on=["source_model", "source_run"],
    how="left",
    validate="1:1",
).sort("source_model", "source_run")
# reapply_agreement.filter(pl.col.agreement_pct.is_not_null())

reapply_agreement.group_by("source_model").agg(
    pl.col("agreement_pct").mean().alias("mean_agreement_pct"),
    pl.col("agreement_pct").std().alias("sd_agreement_pct"),
    pl.col("krippendorff_alpha").mean().alias("mean_alpha"),
    pl.col("krippendorff_alpha").std().alias("sd_alpha"),
)
```
